Importing the Dependencies

```python
import numpy as np
import pandas as pd
import sklearn.datasets
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

Data Collection & Processing

```python
# loading the data from the sklearn
# 0 = Malignant(Cancerous) & 1 = Benign(Non-Cancerous)
breast_cancer_dataset = sklearn.datasets.load_breast_cancer()
```

```python
print(breast_cancer_dataset)
```

```python
# loading the data to a pandas dataset
data_frame = pd.DataFrame(breast_cancer_dataset.data, columns=breast_cancer_dataset.feature_names)
```

```python

# Displaying the first 5 rows of the updated DataFrame
#print(data_frame.iloc[1])
data_frame.head()
```

```python
data_frame['0/1'] = breast_cancer_dataset.target
```

```python
# number of rows and columns in the datasets
data_frame.shape
```

```python
# Getting some information about the data
data_frame.info()
```

```python
# checking the missing values if there are 0/1!
data_frame.isnull().sum()
```

```python
# statistical measures of the data
data_frame.describe()
```

```python
# Distribution of paitents
data_frame['0/1'].value_counts()
```

```python
data_frame.groupby('0/1').mean()
```

Seprating Feature(x) & Target(y)

```python
x = data_frame.drop(columns = '0/1', axis = 1)
y = data_frame['0/1']
```

```python
print(x)
print(y)
```

Splitting the data into Training data & Testing data

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 2)
```

```python
print(x.shape, x_train.shape, x_test.shape)
```

Model Training

Logistic Regression


```python
model = LogisticRegression()
```

```python
# Training Logistic Regression model using x_train & y_train
model.fit(x_train, y_train)
```

Model Evaluation

Accuracy Score

```python
# Accuracy on Training data
x_train_prediction = model.predict(x_train)
training_data_accuracy = accuracy_score(y_train, x_train_prediction)
```

```python
print('Accuracy of the Training Data: ', training_data_accuracy)
```

```python
# Accuracy on Testing data
x_test_prediction = model.predict(x_test)
testing_data_accuracy = accuracy_score(y_test, x_test_prediction)
```

```python
print('Accuracy of the Testing Data: ' , testing_data_accuracy)
```

Building A Prediction System
